# Breakthrough analysis

Run the cells after the simulation has written `column.data` and its component data files.

Simulation `rk3`

```python
from pathlib import Path
import json
import sys

repo_root = Path("/Users/youriran/pusan/ruptura").resolve()
run_dir = Path("/Users/youriran/pusan/ruptura/benchmark/marx-403030_298_1e6_2e-1/grid-10/dt-1e-2/dispersion-0/rk3").resolve()
if repo_root.is_dir() and str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

simulation_json = run_dir / 'simulation.json'
data_dir = run_dir
simulation_config = json.loads(simulation_json.read_text())

from IPython.display import display
simulation_config.get('SimulationType')
```

## Load the breakthrough results

```python
from ruptura.plot_breakthrough import BreakthroughPlotly

plotter = BreakthroughPlotly.from_simulation_json(
    simulation_json,
    data_dir=data_dir,
)
plotter
```

## Breakthrough curves

```python
display(plotter.breakthrough(
    x_units='min',
    y_units='normalized concentration',
    include_carrier_gas=False,
    show_markers=True,
))
```

## Temperature history

```python
try:
    display(plotter.temperature_triplet_time(
        x_units='min',
        y_units='kelvin',
        show_markers=True,
    ))
except Exception as exc:
    print(f'Temperature plot unavailable: {exc}')
```

## Interactive column explorer

```python
explorer = plotter.explorer()
explorer.display()
```
